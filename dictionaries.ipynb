```python
 
student = {'name': 'John', 'age': 25, 'city': 'New York'}
print(student)
```

```python
## accessing dictionary elements
print(student ['city'])
print(student['age'])

## accessing dictionary elements using get() method
print(student.get('name'))
print(student.get('age'))
print(student.get('grade',"not available"))
```

```python
##modifying dictionary elements
## dictionary are mutable, so you can add, update or delete elements 
print(student)
```

```python
student["age"]=26 ## updated a value for key
print(student)
student["grade"]="A" ## added a new key and value
print(student)
```

```python
del student["grade"] ## deleted a key and value
print(student)
```

```python
## dictionary methods

keys = student.keys() ##get all the keys 
print(keys)
values = student.values()## get all values
print(values)

items = student.items() ## get all key value pairs
print(items)
```

```python
## shallow copy 
student_copy=student 
print(student_copy)
print(student)
```

```python
student["name"]="john2"
print(student_copy)
print(student)
```

```python
## for shallow copy, if we change the original dictionary, it will also change the copied dictionary.
student_copy1=student.copy() ## creating a shallow copy using copy() method
print(student_copy1)
print(student)
```

```python
student_copy1["name"]="john3"
print(student_copy1)
print(student)
```

```python
### iterating over dictionaries 
## you can use loops to iterate over the dictionaries , keys , values or items 
for keys in student.keys():
    print(keys)
```

```python
for values in student.values():
    print(values)
```

```python
## iterate over key value paris 
for key, values in student.items():
    print(f"{key}: {values}")
```

```python
## nested disctionaries obviously the dictionaries inside a dictionary
students={
    "student1" : {"name": "John", "age": 32, "city": "New York"},
    "student2" : {"name": "Marie", "age": 22, "city": "Boston"}
}
print(students)
```

```python
##access nested dictionaries elements 
print(students["student1"]["name"])
print(students["student2"]["city"])
```

```python
student.items
```

```python
##iterating over nested dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info.items():
        print(f"  {key}: {value}")
```

```python
## dictionary comprehension
squares = {x:x**2 for x in range(5)}
print(squares)
```
