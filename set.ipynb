```python
## create a set 
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set([1,2,3,4,5,4,2,3])
print(my_empty_set)
```

```python
##basic set operations
my_set.add(6)
print(my_set)
my_set.add(6) 
print(my_set) 
## adding duplicate value will not change the set
```

```python
##remove an element from the set
my_set.remove(4)
print(my_set)
```

```python
my_set.discard(11) 
## discard will not throw an error if the element is not present in the set
print(my_set)
```

```python
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
##clear all the elements from the set
my_set.clear()
print(my_set)
```

```python
## set membership test
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)
print(6 in my_set)
```

```python
##mathematical operations on sets
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

## Union of sets
union_set = set1.union(set2)
print(union_set)

##intersection of sets
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

print(set1.difference(set2))
##the uncommon elements in set1 and set2 can be found using the symmetric_difference() method. It returns a new set with elements that are in either of the sets, but not in both.
set1.symmetric_difference(set2)
```

```python
##sets methods
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5}

##is subset
print(set1.issubset(set2))
print(set2.issubset(set1))
print(set1.issuperset(set2))
```
