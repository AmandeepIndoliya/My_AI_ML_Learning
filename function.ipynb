```python
## keywords arguments 
## all the arguments will be in form of key value arguments 
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
###### examples of functions
## 1st temperature conversion
def convert_temperature(temp,unit):
    """ this function converts temperature between celsius and fahrenheit"""
    if unit=='C':
        return temp * 9/5 + 32 ## celsius to farenheit 
    elif unit=="F":
        return(temp-32)*5/9 ## fahrenheit to celsius 
    else:
        return None


print(convert_temperature(25,'C'))
print(convert_temperature(77,'F'))

    

```

```python
## 2nd password strenght checker 

def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False 
    if not any (char in '!@#$%^&*()' for char in password):
        return False 
    return True 

print(is_strong_password("amandeep"))

print(is_strong_password("J@a8J@a8"))
    
    
      
```

```python
## calculate the total cost of items in a shopping cart 

def sum_of_cart_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+= item['price']* item['quantity']
    return total_cost

cart=[
    {'name':'apple', 'price':0.5,'quantity':4},
    {'name' : 'banana','price':0.5,'quantity':4},
    {'name': 'guava','price':0.7,'quantity': 5}

]
total_cost=sum_of_cart_cost(cart)
print(total_cost)



```

```python
## checking if pelindrome 
def is_palindrome(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print(is_palindrome("a man a plan a canal panama"))
print(is_palindrome("hello"))
```

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))
```
